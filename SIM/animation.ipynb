```python
import matplotlib.pyplot as plt
import numpy as np

import matplotlib.animation as animation

fig, ax = plt.subplots(dpi=300)
ax.set_xlim([0, 10])

scat = ax.scatter(1, 0)
x = np.linspace(0, 10, 100)


anim_len = list(range(len(x)))

pause_len = 30

full_seq = [0] * pause_len + anim_len + [len(x) - 1] * pause_len

def animate(i):
    scat.set_offsets((x[i], 0))
    return (scat,)


ani = animation.FuncAnimation(fig, animate, repeat=True, frames=full_seq, interval=50)

writer = animation.PillowWriter(fps=15,
                                metadata=dict(artist='Fritz Ali Agildere'),
                                bitrate=1800)
ani.save('scatter.gif', writer=writer)

plt.close()
```
