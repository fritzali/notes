## 1. Applications, History, and Core Paradigms

### Applications to Astrophysics and Astrobiology

Machine learning (ML) has become indispensable in processing the massive streams of data generated by modern astronomical surveys (e.g., Vera C. Rubin Observatory, James Webb Space Telescope, Kepler, TESS).

* **Exoplanet Detection and Characterization:** Time-series analysis of stellar light curves using deep learning to isolate shallow transits and radial velocity variations. This is a foundational step in astrobiology for locating habitable zone planets.
* **Biosignature Identification:** Classification of atmospheric transmission spectra to detect chemical non-equilibrium markers (e.g., simultaneous presence of $CH_4$ and $O_2$).
* **Galaxy Morphology and Evolution:** Structural classification of billions of distant galaxies from raw imaging data using computer vision.
* **Cosmic Microwave Background (CMB) Anomalies:** Unsupervised anomaly detection to isolate non-Gaussianities or instrumental artifacts in CMB maps.

### History of Machine Learning: The Method of Least Squares

The mathematical ancestry of machine learning traces back to the early 19th century with Carl Friedrich Gauss and Adrien-Marie Legendre's formulation of the **Method of Least Squares** to solve problems in celestial mechanics (specifically, predicting the orbit of the asteroid Ceres).

Given an overdetermined system of linear equations:

$$\mathbf{y} = \mathbf{X}\mathbf{w} + \boldsymbol{\epsilon}$$

Where $\mathbf{y} \in \mathbb{R}^n$ represents the observations, $\mathbf{X} \in \mathbb{R}^{n \times d}$ is the design matrix, $\mathbf{w} \in \mathbb{R}^d$ is the parameter vector, and $\boldsymbol{\epsilon} \in \mathbb{R}^n$ is the residual error. The optimization objective minimizes the sum of squared residuals:

$$S(\mathbf{w}) = \|\mathbf{y} - \mathbf{X}\mathbf{w}\|^2_2 = (\mathbf{y} - \mathbf{X}\mathbf{w})^T(\mathbf{y} - \mathbf{X}\mathbf{w})$$

To find the optimal parameter vector $\hat{\mathbf{w}}$, we take the gradient with respect to $\mathbf{w}$ and set it to zero:

$$\nabla_{\mathbf{w}} S(\mathbf{w}) = -2\mathbf{X}^T\mathbf{y} + 2\mathbf{X}^T\mathbf{X}\mathbf{w} = \mathbf{0}$$

$$\mathbf{X}^T\mathbf{X}\mathbf{w} = \mathbf{X}^T\mathbf{y}$$

Assuming $\mathbf{X}^T\mathbf{X}$ is non-singular, the analytical solution yields the **Normal Equations**:

$$\hat{\mathbf{w}} = (\mathbf{X}^T\mathbf{X})^{-1}\mathbf{X}^T\mathbf{y}$$

### Core Paradigms: Supervised vs. Unsupervised Learning

* **Supervised Learning:** The dataset contains paired examples $\mathcal{D} = \{(\mathbf{x}_i, y_i)\}_{i=1}^n$. The goal is to learn a mapping $f: \mathcal{X} \to \mathcal{Y}$.
* **Classification:** $\mathcal{Y}$ is a discrete set of categorical labels (e.g., Classifying a star as a *Main Sequence* vs. *Giant*).
* **Regression:** $\mathcal{Y}$ is continuous (e.g., Estimating the metallicity $\text{[Fe/H]}$ of a star).


* **Unsupervised Learning:** The dataset contains only features $\mathcal{D} = \{\mathbf{x}_i\}_{i=1}^n$. The goal is to uncover latent structures, probability densities, or low-dimensional representations within the data (e.g., Clustering stellar populations in Gaia kinematic data).

## 2. Statistical Foundations and Model Selection

### Probability Fundamentals: PDF and CDF

Let $X$ be a continuous random variable.

* **Probability Density Function (PDF), $f_X(x)$:** A function that describes the relative likelihood for this random variable to take on a given value. It satisfies:

$$\forall x \in \mathbb{R}, \, f_X(x) \ge 0 \quad \text{and} \quad \int_{-\infty}^{\infty} f_X(x) \, dx = 1$$

* **Cumulative Distribution Function (CDF), $F_X(x)$:** The probability that $X$ will take a value less than or equal to $x$:

$$F_X(x) = P(X \le x) = \int_{-\infty}^{x} f_X(t) \, dt$$

### Statistical Moments

Moments characterize the shape of a probability distribution. The $n$-th central moment of a random variable $X$ with mean $\mu$ is defined as $\mu_n = \mathbb{E}[(X - \mu)^n]$.

| Moment | Name | Mathematical Formula | Physical/Geometric Interpretation |
| --- | --- | --- | --- |
| 1st Raw | **Mean ($\mu$)** | $\mathbb{E}[X] = \int_{-\infty}^{\infty} x f(x) \, dx$ | Center of mass of the distribution. |
| 2nd Central | **Variance ($\sigma^2$)** | $\mathbb{E}[(X - \mu)^2] = \int_{-\infty}^{\infty} (x - \mu)^2 f(x) \, dx$ | Spread or dispersion around the mean. |
| 3rd Standardized | **Skewness ($\gamma_1$)** | $\frac{\mu_3}{\sigma^3} = \frac{\mathbb{E}[(X - \mu)^3]}{\left(\mathbb{E}[(X - \mu)^2]\right)^{3/2}}$ | Measure of asymmetry. $\gamma_1 > 0$ indicates a long right tail. |
| 4th Standardized | **Kurtosis ($\beta_2$)** | $\frac{\mu_4}{\sigma^4} = \frac{\mathbb{E}[(X - \mu)^4]}{\left(\mathbb{E}[(X - \mu)^2]\right)^2}$ | Measure of tailedness/outlier propensity (peakedness). |

### Frequentist vs. Bayesian Statistics

* **Frequentist Paradigm:** Parameters $\boldsymbol{\theta}$ are fixed, immutable, but unknown constants. Probability represents the limiting relative frequency of an event in an infinite sequence of identical, repeatable trials.
* **Bayesian Paradigm:** Parameters $\boldsymbol{\theta}$ are random variables assigned a probability distribution capturing uncertainty or degree of belief. It updates knowledge based on data via **Bayes' Theorem**:

$$P(\boldsymbol{\theta} | \mathcal{D}) = \frac{P(\mathcal{D} | \boldsymbol{\theta}) P(\boldsymbol{\theta})}{P(\mathcal{D})}$$

Where:

* $P(\boldsymbol{\theta} | \mathcal{D})$ is the **Posterior** distribution.
* $P(\mathcal{D} | \boldsymbol{\theta})$ is the **Likelihood** function $L(\boldsymbol{\theta}; \mathcal{D})$.
* $P(\boldsymbol{\theta})$ is the **Prior** distribution (incorporating physical constraints, e.g., stellar mass must be positive).
* $P(\mathcal{D}) = \int_{\boldsymbol{\theta}} P(\mathcal{D} | \boldsymbol{\theta}) P(\boldsymbol{\theta}) \, d\boldsymbol{\theta}$ is the **Marginal Likelihood** or **Evidence**.

### Point Estimators and Maximum Likelihood Estimation (MLE)

A point estimator $\hat{\boldsymbol{\theta}}$ is a statistic used to approximate a population parameter $\boldsymbol{\theta}$.

Given independent and identically distributed (i.i.d.) observations $\mathcal{D} = \{\mathbf{x}_1, \dots, \mathbf{x}_n\}$ drawn from a distribution $f(\mathbf{x}; \boldsymbol{\theta})$, the Likelihood function is:

$$L(\boldsymbol{\theta}; \mathcal{D}) = \prod_{i=1}^n f(\mathbf{x}_i; \boldsymbol{\theta})$$

Because products are computationally unstable and difficult to differentiate, we maximize the **Log-Likelihood**:

$$\ell(\boldsymbol{\theta}; \mathcal{D}) = \ln L(\boldsymbol{\theta}; \mathcal{D}) = \sum_{i=1}^n \ln f(\mathbf{x}_i; \boldsymbol{\theta})$$

The Maximum Likelihood Estimator is defined as:

$$\hat{\boldsymbol{\theta}}_{\text{MLE}} = \arg\max_{\boldsymbol{\theta}} \sum_{i=1}^n \ln f(\mathbf{x}_i; \boldsymbol{\theta})$$

### Bias, Variance, and the Bias-Variance Tradeoff

Let $\hat{f}(\mathbf{x})$ be an estimator of the true underlying function $f(\mathbf{x})$ evaluated at a target point $\mathbf{x}$, trained on data $\mathcal{D}$. Assuming a noisy target $y = f(\mathbf{x}) + \epsilon$ where $\mathbb{E}[\epsilon] = 0$ and $\text{Var}(\epsilon) = \sigma^2$, the expected squared generalization error can be decomposed as:

$$\mathbb{E}_{\mathcal{D}, \epsilon}\left[(y - \hat{f}(\mathbf{x}))^2\right] = \text{Bias}\left[\hat{f}(\mathbf{x})\right]^2 + \text{Var}\left[\hat{f}(\mathbf{x})\right] + \sigma^2$$

Where:

$$\text{Bias}\left[\hat{f}(\mathbf{x})\right] = \mathbb{E}_{\mathcal{D}}\left[\hat{f}(\mathbf{x})\right] - f(\mathbf{x})$$

$$\text{Var}\left[\hat{f}(\mathbf{x})\right] = \mathbb{E}_{\mathcal{D}}\left[\left(\hat{f}(\mathbf{x}) - \mathbb{E}_{\mathcal{D}}\left[\hat{f}(\mathbf{x})\right]\right)^2\right]$$

* $\sigma^2$ is the **irreducible error** (intrinsic noise in the physical measurement).
* **High Bias** models underfit the training data due to rigid assumptions.
* **High Variance** models overfit by modeling stochastic noise in the training set.

### Regularization, Hyperparameters, and Validation

To control the variance of a model, mathematical penalties are introduced to enforce smoothness or sparsity.

* **$L_2$ Regularization (Ridge Regression):** Adds a penalty proportional to the squared Euclidean norm of the weight vector:

$$\mathcal{L}_{\text{Ridge}}(\mathbf{w}) = \|\mathbf{y} - \mathbf{X}\mathbf{w}\|^2_2 + \lambda \|\mathbf{w}\|^2_2$$

* **$L_1$ Regularization (Lasso Regression):** Adds a penalty proportional to the absolute Manhattan norm of the weight vector, forcing coefficients to exactly zero (feature selection):

$$\mathcal{L}_{\text{Lasso}}(\mathbf{w}) = \|\mathbf{y} - \mathbf{X}\mathbf{w}\|^2_2 + \lambda \|\mathbf{w}\|_1$$

Here, $\lambda \in [0, \infty)$ is a **hyperparameter**—a parameter governing the learning process that cannot be directly optimized via gradient descent on the training data. Hyperparameters must be selected using validation strategies:

* **Holdout Validation:** Splitting data into explicit Training, Validation, and Test partitions.
* **$K$-Fold Cross-Validation:** Partitioning the dataset into $K$ equal-sized subsets. The model is trained on $K-1$ folds and validated on the remaining fold, repeating this sequence $K$ times to yield an aggregate generalization metric.

### The No Free Lunch Theorem

Formulated by Wolpert and Macready, the **No Free Lunch (NFL) Theorem** states that when performance is averaged over all possible data distributions, no single learning algorithm outperforms any other. Formally, for any two algorithms $A_1$ and $A_2$:

$$\sum_{\mathcal{P}} P(e | \mathcal{P}, A_1) = \sum_{\mathcal{P}} P(e | \mathcal{P}, A_2)$$

Where $\mathcal{P}$ is the data-generating distribution and $e$ is the generalization error. This underscores the necessity of tailoring specific inductive biases (e.g., convolutional structures for telescope images) to match the physical constraints of the problem.

## 3. Classical Machine Learning Algorithms

### $k$-Nearest Neighbors ($k\text{NN}$) and the Curse of Dimensionality

$k\text{NN}$ is a non-parametric, instance-based method. Given a query point $\mathbf{x}_0$, it identifies the set $\mathcal{N}_0$ containing the $k$ closest points in the training set based on a distance metric, such as the Euclidean distance:

$$d(\mathbf{x}, \mathbf{z}) = \|\mathbf{x} - \mathbf{z}\|_2 = \sqrt{\sum_{j=1}^d (x_j - z_j)^2}$$

The prediction is computed via majority vote (classification) or averaging (regression):

$$\hat{y}_0 = \frac{1}{k} \sum_{i \in \mathcal{N}_0} y_i$$

#### The Curse of Dimensionality

As the dimensionality $d \to \infty$, the ratio between the distance to the nearest neighbor and the distance to the furthest neighbor rapidly approaches 1:

$$\lim_{d \to \infty} \frac{\text{dist}_{\max} - \text{dist}_{\min}}{\text{dist}_{\min}} = 0$$

Consequently, in high dimensions, all points become nearly equidistant, rendering distance-based metrics functionally uniform and degrading $k\text{NN}$'s performance unless geometric dimensionality reduction is performed.

### Naive Bayes

Naive Bayes classifiers are generative models based on Bayes' theorem, underpinned by the **strong independence assumption** that features are conditionally independent given the class label.

$$P(C_k | \mathbf{x}) = \frac{P(C_k) P(\mathbf{x} | C_k)}{P(\mathbf{x})} \propto P(C_k) \prod_{j=1}^d P(x_j | C_k)$$

The decision rule selects the class maximizing the posterior probability:

$$\hat{y} = \arg\max_{k \in \{1, \dots, K\}} P(C_k) \prod_{j=1}^d P(x_j | C_k)$$

#### Discrete Naive Bayes

For categorical features (e.g., presence or absence of specific emission lines), $P(x_j | C_k)$ is modeled via Multinomial or Bernoulli distributions, often smoothed using Laplace smoothing to avoid zero probabilities:

$$P(x_j = v | C_k) = \frac{N_{k, j, v} + \alpha}{N_k + \alpha V}$$

#### Gaussian Naive Bayes

For continuous features (e.g., planetary radius, stellar temperature), the likelihood of the features is modeled as a normal distribution:

$$P(x_j | C_k) = \frac{1}{\sqrt{2\pi\sigma_{k,j}^2}} \exp\left( -\frac{(x_j - \mu_{k,j})^2}{2\sigma_{k,j}^2} \right)$$

### Support Vector Machines (SVM)

The goal of a linear SVM is to find a separating hyperplane $\mathbf{w}^T\mathbf{x} + b = 0$ that maximizes the functional margin to the closest training points (support vectors).

For a linearly separable dataset $\mathcal{D} = \{(\mathbf{x}_i, y_i)\}_{i=1}^n$, where $y_i \in \{-1, +1\}$, the optimization problem is formulated as:

$$\min_{\mathbf{w}, b} \frac{1}{2}\|\mathbf{w}\|^2_2 \quad \text{subject to} \quad y_i(\mathbf{w}^T\mathbf{x}_i + b) \ge 1, \quad \forall i$$

#### The Primal and Dual Lagrangian

To solve this constrained optimization problem, we introduce Lagrange multipliers $\alpha_i \ge 0$:

$$\mathcal{L}(\mathbf{w}, b, \boldsymbol{\alpha}) = \frac{1}{2}\|\mathbf{w}\|^2_2 - \sum_{i=1}^n \alpha_i \left[ y_i(\mathbf{w}^T\mathbf{x}_i + b) - 1 \right]$$

Taking the partial derivatives with respect to the primal variables $\mathbf{w}$ and $b$, and setting them to zero:

$$\nabla_{\mathbf{w}}\mathcal{L} = \mathbf{w} - \sum_{i=1}^n \alpha_i y_i \mathbf{x}_i = \mathbf{0} \implies \mathbf{w} = \sum_{i=1}^n \alpha_i y_i \mathbf{x}_i$$

$$\frac{\partial \mathcal{L}}{\partial b} = -\sum_{i=1}^n \alpha_i y_i = 0$$

Substituting these back into the primal Lagrangian yields the **Wolfe Dual Problem**:

$$\max_{\boldsymbol{\alpha}} \sum_{i=1}^n \alpha_i - \frac{1}{2} \sum_{i=1}^n \sum_{j=1}^n \alpha_i \alpha_j y_i y_j (\mathbf{x}_i^T \mathbf{x}_j)$$

$$\text{subject to} \quad \alpha_i \ge 0 \quad \forall i, \quad \text{and} \quad \sum_{i=1}^n \alpha_i y_i = 0$$

#### The Kernel Trick

If the data distribution is non-linearly separable, the input space can be mapped to a higher-dimensional feature space $\Phi(\mathbf{x})$. Since the dual formulation depends only on the dot product $\mathbf{x}_i^T \mathbf{x}_j$, we can replace it with a **Kernel Function** $K(\mathbf{x}_i, \mathbf{x}_j) = \langle \Phi(\mathbf{x}_i), \Phi(\mathbf{x}_j) \rangle$, circumventing the explicit computation of $\Phi(\mathbf{x})$:

$$W(\boldsymbol{\alpha}) = \sum_{i=1}^n \alpha_i - \frac{1}{2} \sum_{i=1}^n \sum_{j=1}^n \alpha_i \alpha_j y_i y_j K(\mathbf{x}_i, \mathbf{x}_j)$$

Common kernels include the Radial Basis Function (RBF):

$$K(\mathbf{x}_i, \mathbf{x}_j) = \exp\left( -\gamma \|\mathbf{x}_i - \mathbf{x}_j\|^2_2 \right)$$

#### Probabilistic Outputs (Platt Scaling)

Standard SVMs output uncalibrated distances to the hyperplane $f(\mathbf{x}) = \mathbf{w}^T\Phi(\mathbf{x}) + b$. To obtain calibrated probabilities for non-linearly separable class distributions, **Platt Scaling** passes the scores through a sigmoid function:

$$P(y=1 | \mathbf{x}) = \frac{1}{1 + \exp\left(A \cdot f(\mathbf{x}) + B\right)}$$

Where $A$ and $B$ are scalar parameters optimized via maximum likelihood estimation over an independent validation set.

### Decision Trees and Random Forests

#### Decision Trees

Decision trees recursively partition the feature space based on splitting criteria that minimize impurity. Let node $m$ represent a subset of data $\mathcal{Q}_m$.

* **Gini Impurity (for Classification):** Measures the probability of a random element being misclassified:

$$H(\mathcal{Q}_m) = 1 - \sum_{k=1}^K p_{mk}^2$$

Where $p_{mk} = \frac{1}{N_m} \sum_{\mathbf{x}_i \in \mathcal{Q}_m} \mathbb{I}(y_i = k)$.

* **Shannon Entropy:** Measures information content and uncertainty:

$$H(\mathcal{Q}_m) = - \sum_{k=1}^K p_{mk} \log_2 p_{mk}$$

* **Mean Squared Error (for Regression):**

$$H(\mathcal{Q}_m) = \frac{1}{N_m} \sum_{\mathbf{x}_i \in \mathcal{Q}_m} (y_i - \bar{y}_m)^2, \quad \text{where } \bar{y}_m = \frac{1}{N_m} \sum_{\mathbf{x}_i \in \mathcal{Q}_m} y_i$$

A split partitions $\mathcal{Q}_m$ into left $\mathcal{Q}_m^L$ and right $\mathcal{Q}_m^R$ subsets based on a feature $j$ and threshold $t$. The algorithm maximizes the **Information Gain**:

$$G(\mathcal{Q}_m, j, t) = H(\mathcal{Q}_m) - \frac{N_m^L}{N_m} H(\mathcal{Q}_m^L) - \frac{N_m^R}{N_m} H(\mathcal{Q}_m^R)$$

#### Random Forest Classifier and Regressor

Random Forests are ensemble models that resolve the inherent high-variance nature of single decision trees via two mechanisms of randomness:

1. **Bagging (Bootstrap Aggregating):** Each tree in the ensemble is trained on a dataset $\mathcal{D}_b$ sampled with replacement from the original training set $\mathcal{D}$.
2. **Random Feature Subspaces:** At each node split, only a random subset of features $m \subset d$ (typically $m = \sqrt{d}$ for classification, $m = d/3$ for regression) is evaluated.

The final prediction is aggregated:

$$\hat{y}_{\text{classification}} = \text{mode}\left\{ T_1(\mathbf{x}), T_2(\mathbf{x}), \dots, T_B(\mathbf{x}) \right\}$$

$$\hat{y}_{\text{regression}} = \frac{1}{B} \sum_{b=1}^B T_b(\mathbf{x})$$

## 4. Unsupervised Learning and Dimensionality Reduction

### Principal Component Analysis (PCA)

PCA is a linear transformation technique designed for feature extraction and dimensionality reduction while preserving maximum variance.

Given a design matrix $\mathbf{X} \in \mathbb{R}^{n \times d}$, we center the data such that each column has a mean of zero. The sample covariance matrix $\mathbf{\Sigma} \in \mathbb{R}^{d \times d}$ is computed as:

$$\mathbf{\Sigma} = \frac{1}{n-1} \mathbf{X}^T \mathbf{X}$$

We find the orthogonal projections (principal components) by maximizing the variance of the projected data $\mathbf{X}\mathbf{v}$ subject to $\|\mathbf{v}\|_2^2 = 1$:

$$\max_{\mathbf{v}} \mathbf{v}^T \mathbf{\Sigma} \mathbf{v} \quad \text{subject to} \quad \mathbf{v}^T\mathbf{v} = 1$$

Formulating the Lagrangian:

$$\mathcal{L}(\mathbf{v}, \lambda) = \mathbf{v}^T \mathbf{\Sigma} \mathbf{v} - \lambda(\mathbf{v}^T\mathbf{v} - 1)$$

Taking the derivative with respect to $\mathbf{v}$ and setting it to zero yields the standard **Eigenvalue Equation**:

$$\mathbf{\Sigma}\mathbf{v} = \lambda \mathbf{v}$$

The eigenvectors $\mathbf{v}_j$ represent the principal components, and their corresponding eigenvalues $\lambda_j$ quantify the variance explained along each axis. Dimensionality reduction is achieved by projecting the data onto the top $k$ eigenvectors:

$$\mathbf{X}_{\text{reduced}} = \mathbf{X} \mathbf{V}_k$$

### Autoencoders (AE)

An Autoencoder is an unsupervised neural network trained to copy its input to its output through a bottleneck layer (latent space). It is mathematically structured into two transformations:

$$\text{Encoder:} \quad \mathbf{h} = \sigma_e(\mathbf{W}_e \mathbf{x} + \mathbf{b}_e)$$

$$\text{Decoder:} \quad \hat{\mathbf{x}} = \sigma_d(\mathbf{W}_d \mathbf{h} + \mathbf{b}_d)$$

Where $\mathbf{x} \in \mathbb{R}^d$, $\mathbf{h} \in \mathbb{R}^m$ (with $m \ll d$), $\mathbf{W}_e, \mathbf{W}_d$ are weight matrices, $\mathbf{b}_e, \mathbf{b}_d$ are bias vectors, and $\sigma$ represents non-linear activation functions. The reconstruction loss objective minimizes:

$$\mathcal{L}_{\text{AE}}(\boldsymbol{\theta}) = \frac{1}{n} \sum_{i=1}^n \|\mathbf{x}_i - \hat{\mathbf{x}}_i\|^2_2$$

### Variational Autoencoders (VAE)

Unlike deterministic autoencoders, a VAE is a directed probabilistic graphical model that establishes a continuous, structured latent space $\mathbf{z}$ suited for generative modeling.

#### Mathematical Background and the ELBO

We assume a prior distribution over the latent variables, typically $p(\mathbf{z}) = \mathcal{N}(\mathbf{0}, \mathbf{I})$. The true posterior distribution $p_\theta(\mathbf{z}|\mathbf{x})$ is analytically intractable. We introduce an inference network $q_\phi(\mathbf{z}|\mathbf{x})$ to approximate it. Using Kullback-Leibler (KL) divergence to measure discrepancy:

$$D_{\text{KL}}\left(q_\phi(\mathbf{z}|\mathbf{x}) \parallel p_\theta(\mathbf{z}|\mathbf{x})\right) = \int q_\phi(\mathbf{z}|\mathbf{x}) \ln \left( \frac{q_\phi(\mathbf{z}|\mathbf{x})}{p_\theta(\mathbf{z}|\mathbf{x})} \right) d\mathbf{z}$$

Applying Bayes' theorem to $p_\theta(\mathbf{z}|\mathbf{x}) = \frac{p_\theta(\mathbf{x},\mathbf{z})}{p_\theta(\mathbf{x})}$ and rearranging terms, we derive the **Evidence Lower Bound (ELBO)**:

$$\ln p_\theta(\mathbf{x}) \ge \mathbb{E}_{q_\phi(\mathbf{z}|\mathbf{x})}\left[ \ln p_\theta(\mathbf{x}|\mathbf{z}) \right] - D_{\text{KL}}\left(q_\phi(\mathbf{z}|\mathbf{x}) \parallel p(\mathbf{z})\right)$$

The optimization objective is to maximize this lower bound, which can be expressed as minimizing the negative ELBO loss:

$$\mathcal{L}_{\text{VAE}}(\theta, \phi; \mathbf{x}) = -\mathbb{E}_{q_\phi(\mathbf{z}|\mathbf{x})}\left[ \ln p_\theta(\mathbf{x}|\mathbf{z}) \right] + D_{\text{KL}}\left(q_\phi(\mathbf{z}|\mathbf{x}) \parallel p(\mathbf{z})\right)$$

* The first term is the **reconstruction loss**.
* The second term is the **KL regularization penalty**, forcing the latent distribution to conform to the prior.

#### The Reparameterization Trick

Because $\mathbf{z}$ is sampled stochastically from $q_\phi(\mathbf{z}|\mathbf{x}) = \mathcal{N}(\boldsymbol{\mu}_\phi(\mathbf{x}), \boldsymbol{\Sigma}_\phi(\mathbf{x}))$, standard backpropagation cannot pass gradients through the random sampling node. To resolve this, the **reparameterization trick** isolates the stochasticity by expressing $\mathbf{z}$ deterministically:

$$\mathbf{z} = \boldsymbol{\mu}_\phi(\mathbf{x}) + \boldsymbol{\sigma}_\phi(\mathbf{x}) \odot \boldsymbol{\epsilon}, \quad \text{where } \boldsymbol{\epsilon} \sim \mathcal{N}(\mathbf{0}, \mathbf{I})$$

Where $\odot$ represents the element-wise (Hadamard) product. Gradients can now flow directly through $\boldsymbol{\mu}_\phi$ and $\boldsymbol{\sigma}_\phi$ with respect to the parameters $\phi$.

## 5. Deep Learning Foundations, Optimization, and Advanced Architectures

### Artificial Neural Networks (ANN): The Neuron and Perceptron

The foundational block of an ANN is the artificial neuron (perceptron). Given an input vector $\mathbf{x} \in \mathbb{R}^d$, it computes a weighted linear combination, adds a bias, and passes the result through an activation function $\sigma(\cdot)$:

$$a = \sigma(z) = \sigma\left( \mathbf{w}^T\mathbf{x} + b \right) = \sigma\left( \sum_{j=1}^d w_j x_j + b \right)$$

#### Common Activation Functions

* **Sigmoid:** $\sigma(z) = \frac{1}{1 + e^{-z}}$
* **Hyperbolic Tangent (tanh):** $\tanh(z) = \frac{e^z - e^{-z}}{e^z + e^{-z}}$
* **Rectified Linear Unit (ReLU):** $\text{ReLU}(z) = \max(0, z)$

### Feed-Forward Networks and the Universal Approximation Theorem

A Feed-Forward Neural Network strings together multiple layers of neurons such that information flows strictly forward without cycles:

$$\mathbf{a}^{(l)} = \sigma\left( \mathbf{W}^{(l)} \mathbf{a}^{(l-1)} + \mathbf{b}^{(l)} \right)$$

#### Universal Approximation Theorem (Cybenko, Hornik)

The theorem states that a standard feed-forward neural network with a single hidden layer containing a finite number of neurons, utilizing a non-linear continuous activation function, can approximate any continuous function on compact subsets of $\mathbb{R}^n$ to any desired degree of accuracy $\epsilon > 0$.

### Cost Functions

To train the parameters $\boldsymbol{\Theta} = \{\mathbf{W}, \mathbf{b}\}$, network performance is quantified via a cost function:

* **Mean Squared Error (MSE) [Regression]:**

$$J(\boldsymbol{\Theta}) = \frac{1}{2n} \sum_{i=1}^n \|\mathbf{y}_i - \hat{\mathbf{y}}_i\|^2_2$$

* **Binary Cross-Entropy (BCE) [Binary Classification]:**

$$J(\boldsymbol{\Theta}) = -\frac{1}{n} \sum_{i=1}^n \left[ y_i \ln \hat{y}_i + (1 - y_i) \ln(1 - \hat{y}_i) \right]$$

### Optimization and Gradient Descent

Optimization updates parameters in the direction of the steepest descent of the cost function:

$$\boldsymbol{\Theta}_{t+1} = \boldsymbol{\Theta}_t - \eta \nabla_{\boldsymbol{\Theta}} J(\boldsymbol{\Theta}_t)$$

Where $\eta > 0$ is the learning rate hyperparameter.

#### Gradient Descent Variants

1. **Simple (Stochastic) Gradient Descent (SGD):** Evaluates gradients and updates weights using a single random training sample (or a small batch) at each step.
2. **Momentum:** Accumulates a velocity vector $\mathbf{v}_t$ scaled by a friction coefficient $\beta \in [0, 1)$ to accelerate descent through shallow regions and dampen oscillations:

$$\mathbf{v}_t = \beta \mathbf{v}_{t-1} + \eta \nabla_{\boldsymbol{\Theta}} J(\boldsymbol{\Theta}_t)$$

$$\boldsymbol{\Theta}_{t+1} = \boldsymbol{\Theta}_t - \mathbf{v}_t$$

3. **Adaptive Optimizers (e.g., Adam):** Tracks running estimates of both the first moment (mean) $\mathbf{m}_t$ and the uncentered second moment (variance) $\mathbf{v}_t$ of the gradients to scale individual learning rates element-wise:

$$\mathbf{m}_t = \beta_1 \mathbf{m}_{t-1} + (1 - \beta_1) \mathbf{g}_t, \quad \mathbf{v}_t = \beta_2 \mathbf{v}_{t-1} + (1 - \beta_2) \mathbf{g}_t^2$$

$$\hat{\mathbf{m}}_t = \frac{\mathbf{m}_t}{1 - \beta_1^t}, \quad \hat{\mathbf{v}}_t = \frac{\mathbf{v}_t}{1 - \beta_2^t}$$

$$\boldsymbol{\Theta}_{t+1} = \boldsymbol{\Theta}_t - \frac{\eta}{\sqrt{\hat{\mathbf{v}}_t} + \epsilon} \hat{\mathbf{m}}_t$$

### Backpropagation

Backpropagation evaluates the analytical gradient of the cost function with respect to the network weights using the mathematical calculus **chain rule**. Let $\mathcal{L}$ be the total loss, $z_j^{(l)}$ be the net input to neuron $j$ in layer $l$, and $a_j^{(l)} = \sigma(z_j^{(l)})$ be its activation output.

We define the error term $\delta_j^{(l)}$ for a given neuron as:

$$\delta_j^{(l)} \equiv \frac{\partial \mathcal{L}}{\partial z_j^{(l)}}$$

For the output layer $L$:

$$\delta_j^{(L)} = \frac{\partial \mathcal{L}}{\partial a_j^{(L)}} \sigma'(z_j^{(L)})$$

For any hidden layer $l$:

$$\delta_j^{(l)} = \left( \sum_{k} \delta_k^{(l+1)} w_{kj}^{(l+1)} \right) \sigma'(z_j^{(l)})$$

Expressing this in matrix notation:

$$\boldsymbol{\delta}^{(l)} = \left( (\mathbf{W}^{(l+1)})^T \boldsymbol{\delta}^{(l+1)} \right) \odot \sigma'(\mathbf{z}^{(l)})$$

The partial derivatives of the cost function with respect to individual weights and biases are then computed as:

$$\frac{\partial \mathcal{L}}{\partial w_{ji}^{(l)}} = a_i^{(l-1)} \delta_j^{(l)} \quad \text{and} \quad \frac{\partial \mathcal{L}}{\partial b_j^{(l)}} = \delta_j^{(l)}$$

### Advanced Architectures

#### Convolutional Neural Networks (CNN)

CNNs are designed for spatially correlated data, such as images of galaxies or stellar spectra, by enforcing two geometric constraints: **local connectivity** and **translation invariance**.

The discrete convolution operation in a 2D image layer is formulated as:

$$S(i,j) = (I * K)(i,j) = \sum_{m} \sum_{n} I(i-m, j-n) K(m,n)$$

Where $I$ is the input image array and $K$ is a learnable shared weight kernel.

* **Pooling Layers (e.g., Max Pooling):** Reduces spatial dimensions by extracting the maximum value within a sliding window, downsampling the feature map while building translation invariance.

#### Recurrent Neural Networks (RNN)

RNNs process sequential time-series data (e.g., stellar light curves) by feeding back output states into the hidden layer to maintain sequential memory.

The hidden state $\mathbf{h}_t$ at time step $t$ is calculated via:

$$\mathbf{h}_t = \tanh\left( \mathbf{W}_{hh} \mathbf{h}_{t-1} + \mathbf{W}_{xh} \mathbf{x}_t + \mathbf{b}_h \right)$$

$$\mathbf{y}_t = \text{softmax}\left(\mathbf{W}_{hy} \mathbf{h}_t + \mathbf{b}_y\right)$$

Standard RNNs often suffer from **vanishing or exploding gradients** during backpropagation through time over long sequences due to continuous matrix multiplications. This is mitigated by advanced architectures like Long Short-Term Memory (LSTM) networks or Gated Recurrent Units (GRUs), which introduce internal gating mechanisms.

#### Transformers

Transformers replace recurrent loops entirely with global **Self-Attention** mechanisms, enabling parallelized execution across long-sequence data contexts.

Given an input matrix of vectors, it is mapped into three internal representations via learnable projection matrices: Queries ($\mathbf{Q}$), Keys ($\mathbf{K}$), and Values ($\mathbf{V}$).

$$\mathbf{Q} = \mathbf{X}\mathbf{W}_Q, \quad \mathbf{K} = \mathbf{X}\mathbf{W}_K, \quad \mathbf{V} = \mathbf{X}\mathbf{W}_V$$

The **Scaled Dot-Product Attention** is defined as:

$$\text{Attention}(\mathbf{Q}, \mathbf{K}, \mathbf{V}) = \text{softmax}\left( \frac{\mathbf{Q}\mathbf{K}^T}{\sqrt{d_k}} \right) \mathbf{V}$$

Where $d_k$ represents the dimensionality of the keys, scaling the dot products to prevent vanishing gradients within the softmax function, which transforms raw vectors to valid probability distributions:

$$\text{softmax}(x_i) = \frac{e^{x_i}}{\sum_{j=1}^N e^{x_j}}$$

To learn diverse contextual features simultaneously, **Multi-Head Attention** splits queries, keys, and values into $h$ distinct heads, processes them in parallel, concatenates the outputs, and projects them through a final linear layer:

$$\text{MultiHead}(\mathbf{Q}, \mathbf{K}, \mathbf{V}) = \text{Concat}(\text{head}_1, \dots, \text{head}_h)\mathbf{W}^O$$

$$\text{where} \quad \text{head}_i = \text{Attention}(\mathbf{Q}\mathbf{W}_i^Q, \mathbf{K}\mathbf{W}_i^K, \mathbf{V}\mathbf{W}_i^V)$$
