## Complex Functions

$$z = |z|e^{i(\varphi + 2k\pi)} \in \mathbb{Z}$$

```python
# Preamble

%config InlineBackend.figure_formats = ['retina']
%config InlineBackend.figure_formats = ['svg']

import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt

plt.rcParams['axes.prop_cycle'] = plt.cycler(color=["olivedrab", "steelblue", "goldenrod", "firebrick"]) 
plt.rcParams['figure.figsize'] = [8, 5]
plt.rcParams['figure.constrained_layout.use'] = True
plt.rcParams['legend.frameon'] = False
plt.rcParams["xtick.minor.visible"] = True
plt.rcParams["ytick.minor.visible"] = True

# Ignore Warnings

import warnings
warnings.filterwarnings('ignore')
```

#### Square Root

$$\sqrt{z\,} = |z|^{1/\kern+1pt2}e^{i(\varphi / \kern+1pt 2 + k\pi)} = |z|^{1/\kern+1pt2}
[\cos(\varphi / \kern+1pt 2 + k\pi) + i \sin(\varphi / \kern+1pt 2 + k\pi)]$$

#### Natural Logarithm

$$\log(z) = \log(|z|) + \log\left(e^{i(\varphi + 2k\pi)}\right) = \log(|z|) + i(\varphi + 2k\pi)$$

```python
# Imaginary Parts

def icRoot1(r, p):
    return r**(1/2) * np.sin(p/2)

def icRoot2(r, p):
    return r**(1/2) * np.sin(p/2 + np.pi)

def icLog(r, p):
    return p

# Real Parts

def rcRoot1(r, p):
    return r**(1/2) * np.sin(p/2)

def rcRoot2(r, p):
    return r**(1/2) * np.sin(p/2 + np.pi)

def rcLog(r, p):
    return np.log(r)

# Create Mesh

r = np.linspace(0, 4, 100)
p = np.linspace(0, 2 * np.pi, 200)

R, P = np.meshgrid(r, p)

X = R * np.cos(P)
Y = R * np.sin(P)

# Compute Values

zR1 = icRoot1(R, P)
zR2 = icRoot2(R, P)

wR1 = rcRoot1(R, P)
wR2 = rcRoot2(R, P)

# Calculate Color

normR = mpl.colors.Normalize(vmin=np.min([wR1, wR2]), vmax=np.max([wR1, wR2]))
colMapR = mpl.cm.viridis
scalMapR = mpl.cm.ScalarMappable(cmap=colMapR, norm=normR)
scalMapR.set_array([])

# Plot Function

fig, ax = plt.subplots(1, 2, figsize=[10, 5], subplot_kw={"projection": "3d"})

ax[0].plot_surface(X, Y, zR1, linewidth=0, antialiased=True, facecolors=scalMapR.to_rgba(wR1))
ax[0].plot_surface(X, Y, zR2, linewidth=0, antialiased=True, facecolors=scalMapR.to_rgba(wR2))

ax[0].view_init(elev=25, azim=35, roll=0)
ax[0].set_proj_type('ortho')

ax[0].set_axis_off()

# Create Mesh

r = np.linspace(0, 4, 250)
p = np.linspace(0, 5 * np.pi, 500)

R, P = np.meshgrid(r, p)

X = R * np.cos(P)
Y = R * np.sin(P)

# Compute Values

zL = icLog(R, P)

wL = rcLog(R + 0.236, P)

# Calculate Color

normL = mpl.colors.Normalize(vmin=np.min(wL), vmax=np.max(wL))
colMapL = mpl.cm.plasma
scalMapL = mpl.cm.ScalarMappable(cmap=colMapL, norm=normL)
scalMapL.set_array([])

# Plot Function

ax[1].plot_surface(X, Y, zL, linewidth=0, antialiased=True, facecolors=scalMapL.to_rgba(wL))

ax[1].view_init(elev=25, azim=25, roll=0)
ax[1].set_proj_type('ortho')

ax[1].set_axis_off()

plt.show()
plt.close()
```

In this representation, complex variables $x + iy$ are mapped according to the above functions with their imaginary part shown in the $z$ direction and the real part encoded in a color map of arbitrary normalization. For the square root, its two leaves are recognizable, requiring a cut to ensure holomorphicity. Similarly, the domain of the logarithm needs to be restricted if analyticity is desired.
